```python
#Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy
import PIL
import requests
```

```python
#Creation of random sample data.
np.random.seed(613)

x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
#Recall

#Create a line plot with both Y variables on the same axes.
fig, ax = plt.subplots(figsize= (7,5))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
```

```python
#Adding a label to each plot object.

fig, ax = plt.subplots(figsize= (7,5))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')
fig.show()
```

```python

fig, ax = plt.subplots(figsize= (7,5))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label = "Person 2")

#Adding elements to the labels.
ax.legend(loc = 'lower right', 
          frameon = True, # Add a frame to the legend
          fontsize = 12, # Font size
          ncol = 2, # Number of columns
          shadow = True #Add shadow border
          )
fig.show()
```

```python
fig, ax = plt.subplots(figsize= (7,5))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label = "Person 2")

#Changing label location
ax.legend(loc = 'upper left', 
          bbox_to_anchor = (1,1)
          )
fig.show()
```

### Adding Annotations

```python
fig, ax = plt.subplots(figsize= (7,5))

#Change to scatter plot
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Coordinates and text annotations
ax.text(10,95, "This is very IMPORTANT!")

fig.show()
```

```python
fig, ax = plt.subplots(figsize= (7,5))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Coordinates and text annotations
ax.text(10,95, "This is very IMPORTANT!",
        ha='center',   #Text alignment
        color = 'red', #Font color
        size = 20 #Font size
        )

fig.show()
```

```python
#Creation of a set of axes
fig, ax = plt.subplots()

ax.axis([0, 10, 0, 10])

#Positioning text annotations
ax.text(1, 5, ". Data: (1,5)", transform=ax.transData)
ax.text(0.5, 0.1, ". Axes: (0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure: ( 0.2, 0.2)", transform=fig.transFigure)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Adding arrows to annotations
ax.annotate('This is IMPORTANT!', xy = (10,95), xytext=(20, 94), arrowprops=dict(facecolor='black'))

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Adding arrows to annotations
ax.annotate('This is IMPORTANT!', xy = (10,95), xytext=(20, 94), arrowprops=dict(arrowstyle = 'wedge', facecolor='hotpink'))

fig.show()

#Reference to possible arrow styles: https://matplotlib.org/3.1.1/api/_as_gen/matplotlib.pyplot.annotate.html
```

### Axis Labels

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Removing tick marks and labels
ax.yaxis.set_major_locator(plt.NullLocator()) #Removes Ticks and labels
ax.xaxis.set_major_formatter(plt.NullFormatter()) #Removes ONLY labels.

fig.show()

```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Setting a maximun number of ticks on the axis X
ax.xaxis.set_major_locator(plt.MaxNLocator(3)) 

fig.show()

```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Setting the tick interval to 5
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) 

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Rotating the axis labels, horizontal alignment and font size.
plt.xticks( rotation = 45, ha='right', fontsize=12)


fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc = 'lower right')

#Modifiying axis title fonts.
font1= {'family':'sans-serif', 'color':'olive'}
plt.xlabel('New Font on Axis', fontsize=20, fontdict=font1)


fig.show()
```
