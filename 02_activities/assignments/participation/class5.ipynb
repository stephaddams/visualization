## Seaborn

```python
#Importing libraries and inicializing seed.
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
```

```python
#Loading the dataset.
tips = sns.load_dataset("tips")
print(tips)
```

```python
#Creation of line plot
sns.lineplot(
    data = tips, # Dataset
    x = 'total_bill',
    y = 'tip'
)
```

```python
# Changing the style of the lineplot:
# Options :
# * whitegrid
# * dark
# * darkgrid

sns.set_style('darkgrid')


sns.lineplot(
    data = tips, # Dataset
    x = 'total_bill',
    y = 'tip'
)
```

```python
fig = plt.subplots(figsize=(10,3))

tipgraph = sns.lineplot( data = tips,
                        x = 'total_bill',
                        y = 'tip')

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill($)',
             ylabel = 'Tip Amount ($)')
```

```python
fig = plt.subplots(figsize=(10,3))

#Personalizing the style of the graph.
tipgraph = sns.lineplot( data = tips,
                        x = 'total_bill',
                        y = 'tip',
                        color = 'hotpink',
                        linestyle = '--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor = 'indigo'
                        )

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill($)',
             ylabel = 'Tip Amount ($)')
```

```python
#Creating now a scatterplot
#Personalizing the style of the graph.
tipgraph = sns.scatterplot( data = tips,
                        x = 'total_bill',
                        y = 'tip',
                        style = 'time', #Using the time variable as style
                        hue = 'day',
                        palette= ['purple', 'hotpink', 'blue', 'green'],
                        )

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill($)',
             ylabel = 'Tip Amount ($)')
```

### Pairplot

## Plotly

```python
import plotly.graph_objects as go #  Graph Objects = GO

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([220, 240, 99, 110, 180])
```

```python
# Creating the figure
graph = go.Figure()

# Creating a bar plot type and assigning the data
graph.add_trace(go.Bar(x=x1, y= y1))
```

```python
# Adding titles

graph.update_layout(
    title = 'Pirate Bounties',
    xaxis_title = 'Pirates',
    yaxis_title = 'Bounty'
    )

graph.show()
```

```python
graph = go.Figure()

# Creating a scatter plot type, assigning the data and personalizing the style
graph.add_trace(go.Scatter( x=x1,
                       y= y1,
                       mode ='markers',
                       marker = dict( size= 15,
                                     color = 'hotpink',
                                     opacity = 1,
                                     line = dict( width = 5, color = 'purple')
                                      )
                       ))

# Definition of layout for the whoel graph
graph.update_layout (
    title = "Interactive Pirate Plot",
    xaxis_title = 'Pirates',
    yaxis_title = 'Bounties',
    width = 500,
    height = 500
)
```

## Wordclouds

```python
from wordcloud import WordCloud

df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
                 on_bad_lines = 'skip')

df

```

```python
# Joining all the text from each row in the column quote into a single string.
text = " ".join( each for each in df.quote)

#Generating the Wordcloud Image
wordcloud = WordCloud(background_color= 'white',
                      colormap = 'inferno',
                      random_state = 613).generate(text)

fig, ax = plt.subplots( figsize = (7,3) )

ax.imshow(wordcloud) # imshow for when adding pictures to matplotlib axes
ax.axis("off") # smooths the image
```

## Venn Diagrams

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set (['apple', 'banana', 'watermelon'])
B = set (["pumpkin", 'blueberry', 'apple', 'key lime'])
```

```python
#Creation of diagram, assigning labels and colors.
diagram = venn2( [ A, B ],
                            set_labels= ('Fruits', 'Pies'),
                            set_colors= ('blue', 'red'),
                            alpha = 0.5)


#Defining each circle:

# Left  - Position: 10
# Everything from Fruits (A) that is not in Pies (B)
diagram.get_label_by_id("10").set_text("\n".join(A - B)) 

# Middle  - Position: 11
# Everything that exist in A and B
diagram.get_label_by_id("11").set_text("\n".join(A & B)) 

# Right  - Position: 01
# Everything from Pies (B) that is not in Fruits (A)
diagram.get_label_by_id("01").set_text("\n".join(B - A)) 


plt.show()

##The red circle is bigger cause it contains more elements.
## If we use venn2_unweighted, circles will be the same size
```
