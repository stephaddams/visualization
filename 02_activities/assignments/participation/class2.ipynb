```python
#Heatmap

#Seaborn library has a built in method for heatmap.

import seaborn as sns
import pandas as pd
import numpy as np

# Create a dataset
df = pd.DataFrame(np.random.random((5,5)), columns=["a","b","c","d","e"])

# Default heatmap
p1 = sns.heatmap(df)
```

```python
#Heatmap
import matplotlib.pyplot as plt

# Create a bigger dataset
df = np.random.randn(30, 30)

# Heatmap with different shades of purple and removing the labels of axe Y.
plt.title('Heatmap')
p1 = sns.heatmap(df, cmap="Purples",  yticklabels=False)
```
