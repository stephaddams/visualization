### Creating different types of plots.

```python
#Importing libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Creation of sampling data
np.random.seed(613)

x = np.arange(50)
y = np.random.randint(0, 100, 50)
```

```python
#Cration of basic scatter plot.
# Figure with one plot size 5inch x 3inch

fig, ax = plt.subplots(figsize = ( 5, 3 ))
ax.scatter(x,y)
```

```python
# Creation of a basic vertical bar chart.

fig, ax = plt.subplots(figsize = (5,3))
ax.bar(x,y)
```

```python
# Creation of a basic horizontal bar chart.

fig, ax = plt.subplots(figsize = (5,3))
ax.barh(x,y)
```

```python
# Creation of a basic line chart.

fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)
```

```python
# Creation of a histogram.
# Histograms counts the number of occurences of each value, thats why it only uses one axe.

fig, ax = plt.subplots(figsize = (5,3))
ax.hist(y)
```

### Plot Labels 
#### Adding axis lablels and titles

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

#Graph Labels
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')

#Property that automatically adjusts subplot parameters to fit neatly within the figure area without overlapping or being clipped
fig.tight_layout()
```

```python
#Fonts dictionaries
font1 = {'family':'sans-serif', 'color':'blue', 'size':20}
font2 = {'family':'monospace', 'color':'green', 'size':14}

fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

#Graph Labels
ax.set_title('Total growth over time', fontdict= font1)
ax.set_ylabel('Total growth', fontdict= font2)
ax.set_xlabel('Years since start', fontdict= font2)

#Property that automatically adjusts subplot parameters to fit neatly within the figure area without overlapping or being clipped
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

#Usig LOC to move the title label.
ax.set_title('Total growth over time', fontdict= font1, loc='left')
ax.set_ylabel('Total growth', fontdict= font2)
ax.set_xlabel('Years since start', fontdict= font2)

fig.tight_layout()
```

### Plot apprearance

```python
fig, ax = plt.subplots(figsize = (5,3))

#Modification of appereance of data points.
# Star = *
# Diamonds = D
# Crosses = +
# Hexagons = H

ax.scatter( 
    x,
    y,
    marker='*',
    color= 'indigo'
)
```

```python
fig, ax = plt.subplots(figsize = (5,3))

#Adding a line to connect the datapoints. 
#Changing scatter for plot.

ax.plot( 
    x,
    y,
    marker='*',
    color= 'indigo',
    linestyle = '--',
    linewidth = 2
)
```

```python
fig, ax = plt.subplots(figsize = (5,3))

#Color configuration using HEX Codes.
ax.plot( 
    x,
    y,
    marker='*',
    color= '#7425b9',
    linestyle = '--',
    linewidth = 2,
    markeredgecolor = '#fa9359',
    markerfacecolor = '#000000'
)

#Adding grid lines.
ax.grid()
#To specify the grid only on one axe:
#ax.grid(axis = 'y')
#ax.grid(axis = 'x')
```

```python
fig, ax = plt.subplots(figsize = (5,3))

ax.plot( 
    x,
    y,
    marker='*',
    color= '#7425b9',
    linestyle = '--',
    linewidth = 2,
    markeredgecolor = '#fa9359',
    markerfacecolor = '#000000'
)

#Changing the appearance of the grid.
ax.grid(color='blue', linewidth=2, linestyle='dotted')
```
